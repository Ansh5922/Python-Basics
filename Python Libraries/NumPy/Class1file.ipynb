# Numpy Arrays

NumPy arrays are the main way we will use Numpy throughout the course. Numpy arrays essentially come in two flavors: vectors and matrices. Vectors are strictly 1-d arrays and matrices are 2-d (but you should note a matrix can still have only one row or one column).

Let's begin our introduction by exploring how to create NumPy arrays.

## Creating NumPy Arrays

### From a Python List

We can create an array by directly converting a list or list of lists:

```python
a = [1,2,3,4,5]
arr = np.array(a)
arr
```

```python
arr = np.array([1,2,3,4,5])
arr.dtype
```

```python
arr = np.array(["1",3,4,5,]) # NumPy's type promotion rules require all elements to have a common dtype,
                             # so it promotes everything to strings
arr.dtype
print(arr)
```

Lets create a matrix using list

```python
a = [[1,2,3],[4,5,6],[7,8,9]]
print(a)
```

```python
arr = np.array(a)
arr
```

```python
arr = np.linspace(0,10,5) # Give values at equal distance in an interval
arr
```

```python
np.eye(5,3) #identity matrix
```

```python
np.random.rand(5,4) # random values between 0 - 1
```

```python
np.random.randint(1,100,(3,4)) #random integers [low,high,size]
```

```python
np.random.randn(5,6) # matrix of sixe 5,6 with random values between (-3 - +3)
```

eye → identity matrix
