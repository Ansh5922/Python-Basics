##Fancy Indexing
you can Pick elements from specific positions using a list/array of indices.

```python
import numpy as np
```

```python
arr = np.array([2,4,5,6,7,8])
print(arr[3]) # Access single index
print(arr[[3,5]]) # Access multiple indices
```

You can do this fancy indexing in 2D Matrix aswell

```python
matrix = np.arange(1,19).reshape(3,6)
print(matrix,"\n")

print(matrix[[0,2],[1,4]]) #[[row],[columns]]
                           # 2 = (0,1),
                           # 17 = (2,4)
```

##Conditional Selection
Find values based on conditions or replace them.

```python
arr = np.array([2,4,5,6,7,8])
print(arr>6, '\n')
print(arr[arr>6])
```

```python
print(np.where(arr<6)) # Give indices of elements < 6
print(np.where(arr<6,arr,0)) # Give values < 6 and make rest of the values = 0
print(np.where(arr<6,arr,10)) # Give values < 6 and make rest of the values = 10
```

##Sorting & Ordering
np.sort() → returns sorted array

np.argsort() → returns indices of sorted order

```python
arr = np.array([23,45,21,65,10,18,45])
print(np.sort(arr))
print(np.argsort(arr)) # Returns indices of sorted ordder
```

##Concatenation & Stacking

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.concatenate([a,b]))
print(np.concat([a,b]), '\n')
print(np.stack([a,b]))
```

Vertical & Horizontal Stacking

```python
# matrix
a = np.array([[1,2,3],[6,7,8]])
b = np.array([[4,5,6],[15,16,17]])
print(a,'\n')
print(b)


```

```python
print(np.vstack([a,b]),"\n")
```

```python
print(np.vstack([b,a]))
```

```python
print(np.hstack([a,b]),'\n')
```

```python
print(np.hstack([b,a]),'\n')
```
