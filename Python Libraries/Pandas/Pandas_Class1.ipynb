## 3 Understanding Series & DataFrame
- **Series**: One-dimensional labeled array, like a column in Excel.
- **DataFrame**: Two-dimensional table, similar to an Excel sheet or SQL table.

### Creating a Series from Python Lists or Dictionaries

````python
a = [10,20,30,40]
s = pd.Series(a) # This will give a one-dimensional series
                 # 0,1,2,3 are columns and 0/marks is the only row
print(s)
````

````python
s1 = pd.Series(a,name = "Marks")
s1
````

Dictionary to Series

````python
d = {'susu':90, 'meethu':87, 'PC':100}
s = pd.Series(d, name = 'Marks')
s
````

````python
d = {'susu':90, 'meeth':87, 'PC':100}
s = pd.Series(d, name = 'Marks')
print(s)
````

### Creating a DataFrame

````python
d = {"Name":['susu','meeth','PC'],
     "Marks":[90,87,100],
     "Subject":['Maths','Social','Tecch']}
df = pd.DataFrame(d)

print(df)
````

````python
d = {"Name":['susu','meeth','PC'],
     "Marks":[90,87,100],
     "Subject":['Maths','Social','Tecch']}
df = pd.DataFrame(d)
df
````

### Creating DataFrame from NumPy Arrays

````python
a = np.arange(10,100,10).reshape(3,3)
a
````

````python
d = pd.DataFrame(a, columns = ['A','B','C'])
d
````
