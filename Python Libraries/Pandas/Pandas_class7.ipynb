## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python
import pandas as pd
```

```python
# Dataset 1: Product details
df1 = pd.DataFrame({
    'Product_ID': [1, 2, 3, 4],
    'Name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Price': [1200, 25, 75, 300]
})
display(df1)

# Dataset 2: More products (same columns as df1 → good for vertical stacking)
df2 = pd.DataFrame({
    'Product_ID': [5, 6, 7],
    'Name': ['Webcam', 'Microphone', 'Headphones'],
    'Price': [50, 100, 150]
})
display(df2)

# Dataset 3: Additional info (different columns → good for horizontal stacking)
df3 = pd.DataFrame({
    'Product_ID': [1, 2, 3, 5],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Electronics'],
    'Stock': [100, 200, 150, 80]
})
display(df3)
```

```python
pd.concat([df1, df2], axis=0, ignore_index=True)
```

```python
pd.concat([df1, df3],) # By default axis = 0
```

```python
display(df1)
display(df3)
```

```python
pd.concat([df1, df3], axis=1, ignore_index= False)
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
product = pd.concat([df1, df2], axis=0, ignore_index=True)
display(product)
```

```python
sales = pd.DataFrame({
    "Product_ID": [1, 2, 3, 4,],
    "Quantity sold": [5, 10, 15, 20]
})
display(sales)
```

```python
# On = to choose the columns

pd.merge(product, sales, left_on = "Product_ID", right_on="Product_ID")
```

```python
pd.merge(product, sales, on = "Product_ID", how= "inner") # Include common IDs
```

```python
pd.merge(product, sales, on = "Product_ID", how= "outer") # Inculde every thing
```

```python
pd.merge(product, sales, on = "Product_ID", how= "right") # Focus on right table (sales)
```

```python
pd.merge(product, sales, on = "Product_ID", how= "left") # Focus on left table (product)
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

```python
stock = pd.DataFrame({
    "Stock": [115, 220, 356, 224]
}, index = [1, 2, 3, 5])
display(stock)
```
