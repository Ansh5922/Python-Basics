## 1 Inspecting Data
Quick ways to understand your dataset:
- `head(n)` → first *n* rows (default 5)
- `tail(n)` → last *n* rows
- `info()` → columns, non-null counts, and dtypes
- `shape` → (rows, columns)
- `dtypes` → data type of each column

```python
import pandas as pd

# Create a small DataFrame
a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)
b.head(2) # First 2 rows
```

```python
b.head(10) # 10 > no. of rows in table (4), It gave only 4 rows which are originally available
```

```python
b.tail(2) # Last two rows
```

```python
b.tail(10)
```

```python
b.info()
```

```python
print(b,'\n')
print(b.shape) # (rows, columns)
```

```python
b.dtypes # here dtypes is used because we need to check for multiple rows
```

## 2 Selecting Columns & Rows
Ways to access parts of a DataFrame:
- Single column → `df['col']`
- Multiple columns → `df[['col1','col2']]`
- Label-based selection → `df.loc[row_label, col_label]`
- Integer-index selection → `df.iloc[row_idx, col_idx]`
Tip: `:` means "all rows/columns" in that dimension.

```python

print(b)
b['Name'] # single column selection
```

```python
print(b)
b[['Name','Score']] # Multiple columns selection with "FANCY INDEXING"
```

```python
# Label-based selection → df.loc[row_label, col_label]
print(b)
b.loc[0]
```

```python
type(b.loc[0]) # Its a series
```

```python
print(b)
b.loc[0,'Name'] # It will select single element in row = 0, column = Name
```

```python
b.loc[0:3,['Name','Age']] #
```

```python
# Index based selection, It will select 0th row
b.iloc[0]
```

## 4 Renaming Columns & Setting an Index
- Rename columns → `df.rename(columns={'old':'new'}, inplace=False)`
- Set a column as index → `df.set_index('column_name', inplace=False)`
- Reset index → `df.reset_index(inplace=False)`
 If you don't use `inplace=True`, remember to assign the result back to a variable.

```python
b
c = b.rename(columns={'Score':'Marks'})
c
```

```python
b.rename(columns = {"Score":"Marks"}, inplace = True)  # inplace = True will make changes to the original DataFrame
                                                       # inplace = False will make changes in the copy, not in original DataFrame
b
```

```python
c = b.reset_index(inplace=False) # Reset the index to make 'Name' a column again
c
```

```python
a = b.reset_index(inplace=False) # Reset the index to make 'Name' a column again
a = a.set_index('Name', inplace=False) # Now 'Name' can be set as the index
a
```

## 5 Basic Attributes Recap
- `df.columns` → names of all columns
- `df.index` → row labels
- `df.values` → NumPy array (data only)
These are handy for quick checks and loops (though vectorized ops are preferred).
