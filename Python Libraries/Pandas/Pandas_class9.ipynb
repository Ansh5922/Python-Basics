## 1. Applying Custom Functions with `apply()` / `map()`
- `apply()` lets you apply a custom function across rows or columns.
- `map()` is for applying functions element-wise to a Series.
- Both are handy for transformations, calculations, and cleaning data.

```python
import pandas as pd

# Create a dictionary with some sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Maths': [85, 78, 92, 65, 70],
    'Science': [90, 82, 88, 75, 68],
    'EVS': [70, 60, 75, 80, 85],
    'History': [75, 85, 68, 90, 72]
}

# Create the DataFrame from the dictionary
marks = pd.DataFrame(data)

# Display the first 5 rows of the DataFrame
display(marks.head())
```

```python
marks['Total_Marks'] = marks[['Maths','Science','EVS','History']].apply(sum, axis=1)
display(marks.head())

# apply(lambda row : row.sum(), axis = 1)
```

```python
col_Mean = marks[['Maths','Science','EVS','History']].apply(lambda col : col.mean(), axis=0)
col_Mean
```

```python
# Applying custom function to adataframe

def get_grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
      return 'B'
    elif marks >= 70:
      return 'C'
    elif marks >= 60:
      return 'D'
    elif marks >= 50:
      return 'E'
    else:
      return 'F'


marks['Grade'] = marks['Total_Marks'].apply(get_grade)
display(marks.head())
```

```python
# map() is for applying functions element-wise to a Series.

def grace(marks):
  marks = marks + 15
  return marks

marks['Maths'] = marks['Maths'].map(grace)
display(marks.head())
```
