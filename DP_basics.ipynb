# Dynamic Programming

### 1. Fibonacci series using Memoization

```python
n = int(input("Enter n: "))
def rec(n, dp):
    if n == 0 or n == 1:
        return n
    if dp[n] != -1:
        return dp[n]
    dp[n] = rec(n-1,dp) + rec(n-2,dp)
    return dp[n]
def fib(n):
    dp = [-1]*(n+1)
    return rec(n, dp)
print(fib(n))

```

### 1. Fibonacci series using Tabulation

```python
n = int(input("Enter n: " ))
if n <= 1:
    print(n)
else:
    dp = [-1]*(n+1)
    dp[0] = 0
    dp[1] = 1
    for i in range(2,n+1):
        dp[i] = dp[i-1] + dp[i-2]
    print(dp[n])
```

### Tabulation using constant space

```python
n = int(input("Enter n: " ))
if n <= 1:
    print(n)
else:
    a = 0
    b = 1
    for i in range(2,n+1):
        c = a+b
        a = b
        b = c
    print(b)
```

### Knapsack Problem

```python
n = 5
W = 6 # capacity
weights = [3,2,4,6,4]
values = [10,5,8,5,3]

def knapsack(i,W,n,weights,values):
    if i == n:
        return 0
    if weights[i] <= W:
        take = values[i] + knapsack(i-1,W - weights[i], n, weights, values)
    not_take = knapsack(i+1, W, n, weights,values)
    return max(take, not_take)
ans = knapsack(0,W,n,weights,values)
print ans
```
