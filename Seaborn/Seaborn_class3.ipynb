# Seaborn – Class 3: Distribution Plots

## What Are Distribution Plots?

Distribution plots show how data values are **spread or concentrated**.

Seaborn offers:
- **`histplot()`** → modern histogram (with optional KDE)
- **`kdeplot()`** → smooth curve of distribution
- **`rugplot()`** → small marks showing individual values  
*(Older `distplot()` is now replaced by `histplot()`.)*


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

## histplot() – Basic Histogram

```python
sns.histplot(tips['total_bill']);
```

```python
sns.histplot(tips['total_bill'], kde=True); # Kernel density estimator- density polugon
```

```python
sns.histplot(tips['total_bill'], bins=60); # Custom bins
```

```python
sns.histplot(tips['total_bill'], bins=60, kde=True, color = "red");
```

## kdeplot() – Kernel Density Estimate

```python
# Hyperparameters of kdeplot
# seaborn.kdeplot(data=None, *, x=None, y=None, hue=None, weights=None, palette=None, hue_order=None, hue_norm=None, color=None, fill=None, multiple='layer', common_norm=True, common_grid=False, cumulative=False, bw_method='scott', bw_adjust=1, warn_singular=True, log_scale=None, levels=10, thresh=0.05, gridsize=200, cut=3, clip=None, legend=True, cbar=False, cbar_ax=None, cbar_kws=None, ax=None, **kwargs)
# kernel density estimation.
```

```python
sns.kdeplot(tips['total_bill'], fill = 'True', color = "green");
```

```python
sns.histplot(tips['total_bill'], bins=60, color = "green", kde = 'True', fill = 'True');
sns.kdeplot(tips['total_bill'], fill = 'True', color = "green");
```

## distplot() (Legacy) vs histplot()

> `distplot()` is deprecated.
> Use `histplot()` or `kdeplot()` instead.


## rugplot() – Small Ticks for Observations

```python
sns.rugplot(tips['total_bill'], height = 2);
```

## Multiple Distributions with hue

```python
sns.histplot(x = tips['total_bill'], hue = tips['sex'], bins = 60, kde = 'True',);
```

Key takeaways

histplot() → flexible histogram (can combine with KDE).

kdeplot() → smooth density curve.

rugplot() → quick look at raw values.

distplot() is deprecated → prefer histplot().

Use hue to compare groups in the same distribution plot.
