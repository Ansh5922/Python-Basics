## 1. Find the smallest element in the array

```python
nums = list(map(int, input("Enter numbers separated by space: ").split()))

# The input() function reads a line from input, converts it to a string, and returns it.
# .split() then splits the string by spaces and store them into a list of strings.
# map(int, ...) applies the int() function to each string in the list, converting them to integers.
# list(...) converts the map object into a list.

min_e = float('+inf')
n = len(nums)
for i in range(n):
  if nums[i] < min_e:
    min_e = nums[i]
print(min_e)
```

## 3. Reverse the array

```python
nums = list(map(int, input().split()))

# We perform swapping using two pointers

n = len(nums)
i = 0
j = n - 1

while i< j:
  nums[i] = nums[i] + nums[j]
  nums[j] = nums[i] - nums[j]
  nums[i] = nums[i] - nums[j]

  i = i + 1
  j -= 1

print(nums)
```

```python
# with similar logic we can reverse the digits from/upto kth index

nums = list(map(int, input().split()))
k = int(input())

n = len(nums)

# From kth term
i = k
j = n - 1
nums1 = nums.copy()

while i< j:
  nums1[i] = nums1[i] + nums1[j]
  nums1[j] = nums1[i] - nums1[j]
  nums1[i] = nums1[i] - nums1[j]

  i = i + 1
  j -= 1

print(nums1)


# Upto kth term
i = 0
j = k
nums2 = nums.copy()

while i< j:
  nums2[i] = nums2[i] + nums2[j]
  nums2[j] = nums2[i] - nums2[j]
  nums2[i] = nums2[i] - nums2[j]

  i = i + 1
  j -= 1

print(nums2)
```

## 5. Count the frequency of each element in the array

## 6. Rotate an array by k positions

```python
# e.g:-
# Array (nums): 10 20 30 40 50 60
# Index (k): 2
# Output = [10, 20, 60, 50, 40, 30]
```

```python
nums = list(map(int, input().split()))
k = int(input())

# Reverse the array 3 times

n = len(nums)
i = 0
j = n - 1

while i< j:
  nums[i] = nums[i] + nums[j]
  nums[j] = nums[i] - nums[j]
  nums[i] = nums[i] - nums[j]

  i = i + 1
  j -= 1

i = k + 1
j = n - 1

while i< j:
  nums[i] = nums[i] + nums[j]
  nums[j] = nums[i] - nums[j]
  nums[i] = nums[i] - nums[j]

  i = i + 1
  j -= 1

i = 0
j = k

while i< j:
  nums[i] = nums[i] + nums[j]
  nums[j] = nums[i] - nums[j]
  nums[i] = nums[i] - nums[j]

  i = i + 1
  j -= 1

print(nums)
```

```python
# Instead we can create a reverse function and call it

nums = list(map(int, input().split()))
k = int(input())

def reverse(nums:list, start: int, end: int) -> list:

  i = start
  j = end

  while i < j:
    nums[i] = nums[i] + nums[j]
    nums[j] = nums[i] - nums[j]
    nums[i] = nums[i] - nums[j]

    i = i + 1
    j -= 1

  return nums

reverse(nums, 0, len(nums)-1)
reverse(nums, k+1, len(nums)-1)
reverse(nums, 0, k)

print(nums)

```

## 9. Find Repeating elements of an array

```python
nums = list(map(int, input().split()))

n = len(nums)
table = {}
for i in range(n):
    if nums[i] not in table:
        table[nums[i]] = 1
    else: table[nums[i]] += 1
print("Unique elements", end= ' = ')
print(table.keys()) # unique elements

for key, value in table.items():
    if value > 1:
        print(f"Element {key} repeats {value} times")
```

## 10. Print the symmetric pairs

```python
nums = list(map(int, input().split()))

# Group consecutive pairs (jump by 2)
pairs = [(nums[i], nums[i+1]) for i in range(0, len(nums), 2)]

print(pairs)

table = {}
sym_pairs = []

for (first, second) in pairs:
    if second in table and table[second] == first:
        sym_pairs.append((first,second))
    else: table[first] = second

print(sym_pairs)
```

## 11. Find 'rank' of given element in an array

## 12. Equilibrium Point

```python
# Eqm Point: A point in array where sum of all the elements to its left is equals to sum of all the elements to its right

nums = list(map(int, input().split()))
def eqm(nums: list):
    n = len(nums)
    rsum = 0
    lsum = 0
    found = False
    for i in range(n):
        rsum += nums[i]
    for i in range(n):
        rsum -= nums[i]
        if rsum == lsum:
            print(f'{nums[i]} is the equilibrium point')
            found = True
        lsum += nums[i]
    if not found:
        return "No eqm point"

print(eqm(nums))
```

## 14. Binary Search: for sorted array
time complexity: 𝑂(log𝑛) → because the search space is halved at each step.

```python
nums = sorted(list(map(int, input().split())))
k = int(input())
n = len(nums)

def bs(nums: list, k:int):
    start = 0
    end = n - 1
    found = False
    while start <= end:
        mid = (end + start) // 2
        if nums[mid] == k:
            print(f'Number found at index {mid}')
            found = True
            return
        elif nums[mid] > k:
            end = mid - 1
        else: start = mid + 1
    if not found: print('Number not found')

bs(nums, k)
    
```

## 15. Subarray

```python
nums1 = [1,2,3]
nums = [1,2,3,4,5,6,7,8,9]

def is_subarray(a: list, b: list):
    n, m = len(a), len(b)
    for i in range(n-m+1):
        if a[i:i+m] == b:
            return True
    return False

print(is_subarray(nums,nums1))

```
