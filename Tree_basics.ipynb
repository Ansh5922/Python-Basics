Preorder Traversal

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = None
root.left.right = Node(4)
root.right.right = Node(5)
root.right.left = Node(6)
    
```

```python
class Preorder:
    ans = []
    def preorder(self, root):
        if root == None:
            return
        self.ans.append(root.val)
        self.preorder(root.left)
        self.preorder(root.right)
    def preorder_traversal(self, root):
        self.ans = []
        self.preorder(root)
        return self.ans
    
p = Preorder()
print(p.preorder_traversal(root))
```

Postorder Traversal

```python
class Postorder:
    ans = []
    def postorder(self, root):
        if root == None:
            return
        self.postorder(root.left)
        self.postorder(root.right)
        self.ans.append(root.val)
    def postorder_traversal(self, root):
        self.ans = []
        self.postorder(root)
        return self.ans
    
p = Postorder()
print(p.postorder_traversal(root))
```

Inorder Traversal

```python
class Inorder:
    ans = []
    def inorder(self, root):
        if root == None:
            return      
        self.inorder(root.left)
        self.ans.append(root.val)
        self.inorder(root.right)
    def inorder_traversal(self, root):
        self.ans = []
        self.inorder(root)
        return self.ans
    
p = Inorder()
print(p.inorder_traversal(root))
```
